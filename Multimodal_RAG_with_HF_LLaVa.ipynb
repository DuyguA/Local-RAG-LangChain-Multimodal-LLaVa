# Load documents

```python
texts = []

images = ["images/image"+str(i)+".jpg" for i in range(1,4)]
captions = ["images/cap"+str(i)+".txt" for i in range(1,4)]

for cname in captions:
  caption = open(cname).read().strip()
  texts.append(caption)
```

# Define embeddings & create the vectorbase

```python
import open_clip

open_clip.list_pretrained()
```

```python
from langchain_experimental.open_clip import OpenCLIPEmbeddings

embedding_function = OpenCLIPEmbeddings(
    model_name="ViT-B-32",
    checkpoint="laion2b_e16",
    load_fn_kwargs={"device": "cuda"}
)
```

```python
from langchain_chroma import Chroma

vectorstore = Chroma(
    collection_name="mini_photos", embedding_function=embedding_function, persist_directory="mini-db")

# Get image URIs with .jpg extension only

# Add images
vectorstore.add_images(uris=images)

# Add documents
vectorstore.add_texts(texts=texts)

# Make retriever
retriever = vectorstore.as_retriever()
```


# Prepare LLaVA

```python
from transformers import AutoProcessor, LlavaForConditionalGeneration
import torch

model_name = "llava-hf/llava-1.5-7b-hf"

model = LlavaForConditionalGeneration.from_pretrained(
    model_name,
    torch_dtype=torch.bfloat16,
    load_in_4bit=True,
    device_map="auto")

processor = AutoProcessor.from_pretrained(model_name)
```

```python
import base64
from PIL import Image
from io import BytesIO


def decode_base64_image(base64_string):
    """Decode a Base64 string into a PIL.Image.Image object."""
    image_data = base64.b64decode(base64_string)
    return Image.open(BytesIO(image_data))


def is_base64(s):
    """Check if a string is Base64 encoded"""
    try:
        return base64.b64encode(base64.b64decode(s)) == s.encode()
    except Exception:
        return False

def reshape_results(docs):
    images = []
    text = []
    for doc in docs:
        doc = doc.page_content  # Extract Document contents
        if is_base64(doc):
            images.append(
                decode_base64_image(doc)
            )  # base64 encoded str
        else:
            text.append(doc)
    return {"images": images, "texts": text}
```

```python
def prompt_func(data_dict):
    formatted_texts = data_dict["texts"][0]
    first_image=None

    if data_dict["images"]:
      first_image = data_dict["images"][0]

    text_message = (
            "As an expert cultural critic and historian, your task is to analyze and interpret images, "
            "considering their historical and cultural significance. Alongside the images, you will be "
            "provided with related text to offer context. Please give a detailed description of the visual elements in the image.\n"
            "Context:\n"
            f"{formatted_texts}"
        )
    conversation = [
      {
      "role": "user",
      "content": [
          {"type": "text", "text": text_message},
          {"type": "image"},
        ],
      },
     ]
    return conversation, first_image
```

```python
query = "Traditional Indian women wear"

results = retriever.invoke(query, k=3)

data_dict = reshape_results(results)
conversation, first_image= prompt_func(data_dict)

final_prompt= processor.apply_chat_template(conversation, add_generation_prompt=True)
```

```python
inputs = processor(images=first_image, text=final_prompt, return_tensors='pt').to(0, torch.float16)

output = model.generate(**inputs, max_new_tokens=200, do_sample=False)
print(processor.decode(output[0][2:], skip_special_tokens=True))
```
